```python
import os
from openai import OpenAI
from dotenv import load_dotenv

# 1. Load the API key from your .env file
load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")

# 2. Redirect the OpenAI library to target Google AI Studio's servers
client = OpenAI(
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

# 3. Define your target model string
MODEL_NAME = "gemma-4-31b-it" # Or "gemma-4-26b-a4b-it"

print(f"Starting standard OpenAI chat completion with: {MODEL_NAME}...\n")

# 4. Construct a message payload history list
messages_history = [
    {
        "role": "system", 
        "content": "You are a helpful, senior software engineering mentor."
    },
    {
        "role": "user", 
        "content": "Hi! Give me 2 quick personal project ideas using Gemma 4."
    }
]

# --- Turn 1 ---
response_1 = client.chat.completions.create(
    model=MODEL_NAME,
    messages=messages_history,
    temperature=0.7
)

# Extract and append the model's text answer to keep the chat loop context alive
gemma_reply_1 = response_1.choices[0].message.content
print(f"Gemma:\n{gemma_reply_1}\n")
messages_history.append({"role": "assistant", "content": gemma_reply_1})
print("-" * 60)

# --- Turn 2 (Testing Historical Context Retention) ---
user_prompt_2 = "Which of those is faster to deploy for a frontend dev?"
print(f"User: {user_prompt_2}\n")
messages_history.append({"role": "user", "content": user_prompt_2})

response_2 = client.chat.completions.create(
    model=MODEL_NAME,
    messages=messages_history,
    temperature=0.7
)

gemma_reply_2 = response_2.choices[0].message.content
print(f"Gemma:\n{gemma_reply_2}\n")
messages_history.append({"role": "assistant", "content": gemma_reply_2})
print("-" * 60)
```
