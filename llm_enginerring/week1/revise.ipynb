How we call llm in short 

```python
# // first we have import some packages which we want 
import os;
from openai import OpenAI;
from IPython.display import Markdown,display;
```

```python
base_url_ollama="http://localhost:11434/v1"
api_key="ollama"
Model="llama3.2"
ollama_client = OpenAI(
    base_url=base_url_ollama,
    api_key=api_key
)
system_prompt ="Your are helpful coding assistant sometime tells joke"
def give_chatmessage(system,user):
    ans = [
            {"role": "system", "content": system},
            {"role": "user", "content": user}
        ]
    return ans
give_chatmessage(system_prompt,"hello")

```

```python
def call_ollama(messages):
    response = ollama_client.chat.completions.create(
         model=Model,
         messages=messages
    )
    result = response.choices[0].message.content
    display(Markdown(result))
call_ollama(give_chatmessage(system_prompt,"hello good morning"));
```
