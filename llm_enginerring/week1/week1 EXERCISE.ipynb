```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'

system_prompt = """
You are a personal AI tutor for Neeraj — a frontend developer at Adobe with solid React experience, 
now systematically learning LLM engineering.

When explaining concepts:
- Assume he knows JavaScript/React well — use those as analogies where helpful
- Assume basic Python familiarity — no need to explain syntax basics
- He understands software engineering fundamentals (APIs, async, data structures)
- He is NOT yet familiar with ML/AI internals — explain those from scratch
- Focus on the WHY not just the WHAT — he learns by understanding the reasoning
- Be direct and concise — skip fluff, get to the point

Response structure:
1. One-line plain English answer first
2. Step-by-step breakdown with a concrete example
3. Why it matters / when you'd use this in practice
4. Common gotcha or edge case worth knowing (if any)
"""
```

```python
# load_dotenv(override=True)

# openai_client = OpenAI()

ollama_client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)
```

```python
# Read a local file and embed its content in the question
def read_file(path):
    with open(path, "r", encoding="utf-8") as f:
        return f.read()

# Change the topic to whatever you want to ask about
PROGRESS_FILE = r"c:\Users\neerajj\Usersneerajjllm_engineering\progress.md"
topic = "Day 3"

progress_content = read_file(PROGRESS_FILE)

question = f"""
Here is my course progress notes:

{progress_content}

Based on the notes above, give me a concise revision summary of {topic} — 
key concepts, why they matter, and anything I should make sure I remember.
"""
```

```python
def ask_llama(question):
    response = ollama_client.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
        ]
    )
    display(Markdown(response.choices[0].message.content))

ask_llama(question)
```
